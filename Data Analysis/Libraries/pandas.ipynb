## Series: 
 ### A panda  series is a one dimensional array-like an object that can hold any data type. It is similar to a column in table.

```python
import pandas as pd
data = [1,2,3,4,5]
series = pd.Series(data)
print("Series \n", series)
```

```python
## Create a Series from Dictionary
data = {'a':1, 'b':2, 'c':3}
series_dict=pd.Series(data)
print(series_dict)
```

```python
data =[36, 38, 42]
index=["Abhi", "Shristy", "Arpit"]
pd.Series(data, index = index)
```

## DataFrame

```python
## Create a DataFrame from a dictionary of List
data={
    'Name': ["Arpit","Shristy","Abhishek"],
    'RollNo': [42, 38, 36],
    'District': ["Bhojpur", "Rohtas", "Patna"]
}
index = [1, 2, 3]
df = pd.DataFrame(data, index = index)
print(df)
print(type(df))
```

```python
import numpy as np
np.array(df)
```

```python
## Create a DataFrame from a list of Dictionaries

data = [
    {'Name': "Arpit", 'RollNo': 42, "District": "Bhojpur"},
    {'Name': "Shristy", 'RollNo': 38, "District": "Rohtas"},
    {'Name': "Abhi", 'RollNo': 36, "District": "Patna"},
]
df = pd.DataFrame(data)
print(df)
print(type(df))
```
